Q2. Basic NumPy array questions

```python
# a) reverse the array
arr = np.array([1, 2, 3, 6, 4, 5])
print('original:', arr)
print('reversed:', arr[::-1])
```

```python
# b) most frequent element and its indices
def most_frequent(arr):
    values, counts = np.unique(arr, return_counts=True)
    top = counts.max()
    for v in values[counts == top]:
        print('element', v, 'occurs', top, 'times at indices', np.where(arr == v)[0])

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

print('x:')
most_frequent(x)
print()
print('y:')
most_frequent(y)
```

Q4. 1D array using linspace

```python
kovid_kumar = np.linspace(10, 100, 25)
print(kovid_kumar)
```

```python
print('dimensions   :', kovid_kumar.ndim)
print('shape        :', kovid_kumar.shape)
print('total elements:', kovid_kumar.size)
print('data type    :', kovid_kumar.dtype)
print('total bytes  :', kovid_kumar.nbytes)
```

```python
# transpose using reshape: a 1D array has no second axis, so turn it into a column
reshaped_kk = kovid_kumar.reshape(25, 1)
print(reshaped_kk.shape)
print(reshaped_kk)
```

```python
# trying the T attribute
print(kovid_kumar.T.shape)
print(np.array_equal(kovid_kumar, kovid_kumar.T))
```

No, we cannot get the same result with the T attribute. For a 1D array, T returns the same array with shape (25,) because there is only one axis to swap. To get a column we need reshape(25, 1), or the array has to be 2D first, e.g. kovid_kumar.reshape(1, 25).T.

Q5. 2D array operations

```python
ucs420_kovid = np.array([[10, 20, 30, 40],
                         [50, 60, 70, 80],
                         [90, 15, 20, 35]])
print(ucs420_kovid)
print('shape:', ucs420_kovid.shape)
```

```python
print('mean   :', np.mean(ucs420_kovid))
print('median :', np.median(ucs420_kovid))
print('max    :', np.max(ucs420_kovid))
print('min    :', np.min(ucs420_kovid))
print('unique elements:', np.unique(ucs420_kovid))
```

```python
# reshape to four rows and three columns
reshaped_ucs420_kovid = ucs420_kovid.reshape(4, 3)
print(reshaped_ucs420_kovid)
print('shape:', reshaped_ucs420_kovid.shape)
```

```python
# resize to two rows and three columns (keeps only the first 6 elements)
resized_ucs420_kovid = np.resize(ucs420_kovid, (2, 3))
print(resized_ucs420_kovid)
print('shape:', resized_ucs420_kovid.shape)
```
