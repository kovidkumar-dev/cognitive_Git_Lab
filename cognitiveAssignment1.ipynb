## Assignment 5.3

WAP to add all prime numbers from 1 to n, where n is given by the user.

```python
# Write your solution here
n = int(input("Enter a number: "))
s = 0
for num in range(2, n + 1):
    f = 0
    for i in range(2, num // 2 + 1):
        if num % i == 0:
            f = 1
            break
    if f == 0:
        s = s + num
print("Sum of primes -->", s)
```

## Assignment 6.1

WAP using a function to add all odd numbers from 1 to n, where n is given by the user.

```python

def AddOdd(n):
    s = 0
    for i in range(1, n + 1):
        if i % 2 != 0:
            s = s + i
    return s

print("AddOdd(10) -->", AddOdd(10))
print("AddOdd(20) -->", AddOdd(20))
```

## Assignment 6.2

WAP using a function to add all prime numbers from 1 to n, where n is given by the user.

```python

def IsPrime(n):
    if n < 2:
        return 0
    for i in range(2, n // 2 + 1):
        if n % i == 0:
            return 0
    return 1

def AddPrimes(n):
    s = 0
    for i in range(1, n + 1):
        if IsPrime(i):
            s = s + i
    return s

print("AddPrimes(20) -->", AddPrimes(20))
print("AddPrimes(50) -->", AddPrimes(50))
```
