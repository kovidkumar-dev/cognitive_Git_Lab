Q1. Sensor readings

```python
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
print(temperature)
```

```python
# a) add 2 degrees to every reading
adjusted = temperature + 2
print(adjusted)
```

```python
# b) celsius to fahrenheit, F = 9/5 * C + 32
fahrenheit = (9 / 5) * temperature + 32
print(fahrenheit)
```

```python
# c) readings greater than 32
hot = temperature[temperature > 32]
print(hot)
```

```python
# d) count of readings above 32
print(np.sum(temperature > 32))
```

e) Vectorization applies an operation to the whole array at once, and the looping happens inside NumPy's compiled C code instead of in slow Python bytecode. That makes it much faster than going through the values one by one, and the code is also shorter and easier to read. Boolean indexing works the same way: the condition creates a True/False mask for all elements together, and the mask picks out the matching values (or counts them with sum) without writing an if statement inside a loop. For large datasets this saves a lot of time.

Q2. Daily steps of four users

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]])
print(steps)
```

```python
# a) total steps
print(np.sum(steps))
```

```python
# b) mean steps
print(np.mean(steps))
```

```python
# c) max and min
print('max:', np.max(steps))
print('min:', np.min(steps))
```

```python
# d) total steps for each day (down the rows)
print(np.sum(steps, axis=0))
```

```python
# e) total steps for each user (across the columns)
print(np.sum(steps, axis=1))
```

```python
# f) position of the maximum value
flat_index = np.argmax(steps)
user, day = np.unravel_index(flat_index, steps.shape)
print('flat index:', flat_index)
print('user index:', user, ' day index:', day)
print('value:', steps[user, day])
```

Indexing starts from 0, so user index 3 is the 4th user and day index 2 is the 3rd day.

Q3. Slicing, copying, reshaping and flattening

```python
# a) create the array
original = np.array([1, 2, 3, 4, 5, 6])
print(original)
```

```python
# b) slice from index 1 to 4
subset = original[1:5]
print(subset)
```

```python
# c) change first element of subset (it is a view, so original changes too)
subset[0] = 999
print('original:', original)
print('subset  :', subset)
```

```python
# d) slice with copy, original stays unchanged
copied = original[1:5].copy()
copied[0] = 500
print('original:', original)
print('copied  :', copied)
```

```python
# e) numbers 1 to 12 as a 3 x 4 matrix
matrix = np.arange(1, 13).reshape(3, 4)
print(matrix)
```

```python
# f) indexing and slicing
print('first row   :', matrix[0])
print('last row    :', matrix[-1])
print('second col  :', matrix[:, 1])
print('rows 1-2, cols 2-3:')
print(matrix[0:2, 1:3])
```

In the last one I took rows 1 to 2 and columns 2 to 3 as normal counting, so in Python indices that is matrix[0:2, 1:3].

```python
# g) flatten and ravel
flat = matrix.flatten()
rav = matrix.ravel()
print('flatten:', flat)
print('ravel  :', rav)
```

```python
# h) change an element in the ravel result
rav[0] = 100
print(rav)
print(matrix)
```

```python
# i) change an element in the flatten result
flat[1] = 200
print(flat)
print(matrix)
```

h) ravel returns a view when it can, so changing it also changed the original matrix (the first element became 100).

i) flatten always returns a copy, so changing it did not affect the original matrix.

```python
# j) details of the matrix (matrix was changed in part h, but the properties are the same)
print('shape:', matrix.shape)
print('ndim :', matrix.ndim)
print('size :', matrix.size)
print('dtype:', matrix.dtype)
```

Q4. Predicting the assistance score with least squares

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]])
y = np.array([40, 65, 30, 85])
```

```python
# a) shape and dimensions of X
print('shape:', X.shape)
print('ndim :', X.ndim)
```

```python
# b) transpose
print(X.T)
print(X.T.shape)
```

b) The transpose swaps rows and columns. X has users as rows and features as columns (4 x 3), so X.T has features as rows and users as columns (3 x 4). Each row of X.T now holds one feature (sleep, activity, stress) for all four users.

```python
# c) X.T @ X
XtX = X.T @ X
print(XtX)
```

```python
# d) inverse of X.T @ X, checking the determinant first so we know it can be inverted
print('determinant:', np.linalg.det(XtX))
XtX_inv = np.linalg.inv(XtX)
print(XtX_inv)
```

The determinant is not zero, so the matrix is invertible and np.linalg.inv() can be used here.

```python
# e) ordinary least squares, beta = (X^T X)^-1 X^T y
beta = np.linalg.inv(X.T @ X) @ X.T @ y
print(beta)
```

```python
# check against numpy's built in least squares solver
print(np.linalg.lstsq(X, y, rcond=None)[0])
```

f) The three coefficients are the weights for sleep hours, activity level and stress level, in that order. Each one tells how much the predicted assistance score changes when that feature goes up by 1 and the other two stay the same.

Sleep is about -0.16, so more sleep lowers the score very slightly. Activity is about 0.14, so a higher activity level raises the score a little. Stress is about 10.06, so it has by far the biggest effect and each extra stress point raises the predicted score by about 10.

The activity sign is a bit unexpected, and this is because there are only 4 users and no intercept term in the model, so the coefficients should not be taken too seriously.

```python
# g) prediction for the new user
new_user = np.array([5, 40, 7])
prediction = new_user @ beta
print('predicted assistance score:', prediction)
```
