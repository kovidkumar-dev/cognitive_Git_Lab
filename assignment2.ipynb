## Q1. Lists
---

**i.** Build list `L` (each digit × 10) and print it

```python
L = [d * 10 for d in digits]
print("Original list L:", L)
```

**ii.** Add two numbers — one via `append()`, one via `insert()`

```python
L.append(100)                 # append() adds an element at the END of the list
print("After append(100):", L)

L.insert(3, 999)               # insert(pos, val) places 999 at index 3, shifting later elements right
print("After insert(3, 999):", L)
```

**iii.** Remove two elements — one via `remove()`, one via `pop()`

```python
L.remove(100)                  # remove() deletes the FIRST occurrence of the given VALUE (100)
print("After remove(100):", L)

L.pop(3)                        # pop(index) deletes the element at that INDEX and returns it (removes 999)
print("After pop(3):", L)
```

**iv.** Sort ascending, then descending

```python
L.sort()                        # sorts in place, ascending
print("Ascending sort:", L)

L.sort(reverse=True)            # sorts in place, descending
print("Descending sort:", L)
```

**v.** Slicing — first three and last three elements

```python
print("First three elements:", L[:3])
print("Last three elements: ", L[-3:])
```

**vi.** List comprehension — elements greater than the average of `L`

```python
avg_L = sum(L) / len(L)
above_avg = [x for x in L if x > avg_L]
print(f"Average of L = {avg_L}")
print("Elements greater than average:", above_avg)
```

## Q2. Tuples
---

Create `scores` tuple using the first 8 values of `L` from Q1

```python
scores = tuple(L[:8])
print("scores tuple:", scores)
```

**i.** Highest score + its index, lowest score + how many times it appears

```python
highest = max(scores)
highest_index = scores.index(highest)
lowest = min(scores)
lowest_count = scores.count(lowest)
print(f"Highest score = {highest} at index {highest_index}")
print(f"Lowest score = {lowest}, appears {lowest_count} time(s)")
```

**ii.** Reverse the tuple, return the result as a list

Tuples are immutable, so they have no in-place `.reverse()` method like lists do;
reversing must instead produce a brand-new tuple/list via slicing `[::-1]`.

```python
reversed_scores = list(scores[::-1])
print("Reversed scores (as list):", reversed_scores)
```

**iii.** Ask the user to input a score; print its first occurrence index or a not-present message

```python
user_score = int(input("Enter a score to search for: "))
if user_score in scores:
    print(f"{user_score} found at index {scores.index(user_score)}")
else:
    print(f"{user_score} is not present in the tuple")
```

**iv.** Attempt to change one element of the tuple directly — capture the error

Tuples are immutable (their contents cannot be changed after creation), so item
assignment raises a `TypeError`; a list would have allowed this assignment freely.

```python
print("Attempting scores[0] = 100 ...")
try:
    scores[0] = 100
except TypeError as e:
    print("Error raised:", e)
```

**v.** Unpack the tuple into a first score, a second score, and all remaining scores

```python
first_score, second_score, *remaining_scores = scores
print("first_score =", first_score)
print("second_score =", second_score)
print("remaining_scores =", remaining_scores)
```

## Q3. Random Numbers
---

Seed the RNG with the roll number so results are reproducible but unique

```python
random.seed(1024170353)
```

**i.** Generate a list of 100 random numbers between 100 and 900 (inclusive)

```python
rand_list = [random.randint(100, 900) for _ in range(100)]
print("Random list (first 20 shown):", rand_list[:20], "...")
```

**ii.** Count all odd numbers

```python
odd_count = len([x for x in rand_list if x % 2 != 0])
print("Count of odd numbers:", odd_count)
```

**iii.** Count all even numbers

```python
even_count = len([x for x in rand_list if x % 2 == 0])
print("Count of even numbers:", even_count)
```

**iv.** Count and list all prime numbers (via list comprehension)

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

primes = [x for x in rand_list if is_prime(x)]
print("Count of prime numbers:", len(primes))
print("Prime numbers found:", primes)
```

**v.** Number that occurs most frequently, and how many times

```python
most_common = max(set(rand_list), key=rand_list.count)
most_common_count = rand_list.count(most_common)
print(f"Most frequent number = {most_common}, occurs {most_common_count} time(s)")
```

## Q4. Sets
---

Build sets `A` (digit × 7) and `B` (digit × 9) from the first 8 raw digits

```python
roll_digits_8 = digits[:8]        # first 8 digits, BEFORE multiplying by 10
A = {d * 7 for d in roll_digits_8}
B = {d * 9 for d in roll_digits_8}
print("Set A (digit x 7):", A)
print("Set B (digit x 9):", B)
```

**vi.** Union of A and B

```python
print("Union A | B:", A | B)
```

**vii.** Intersection of A and B

```python
print("Intersection A & B:", A & B)
```

**viii.** `A - B` and `B - A` separately

`difference()` returns only the elements unique to the FIRST set (one direction),
whereas `symmetric_difference()` returns elements unique to EITHER set combined (both directions).

```python
print("A - B:", A.difference(B))
print("B - A:", B.difference(A))
```

**ix.** Symmetric difference of A and B

```python
print("Symmetric difference A ^ B:", A.symmetric_difference(B))
```

**x.** Subset / superset checks

```python
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))
```

**xi.** Ask the user for a value X, remove it from A using `discard()`

`discard()` does NOT raise an error if the value is absent, unlike `remove()`,
which is safer when you're not sure the value exists in the set.

```python
x_val = int(input("Enter a value to discard from set A: "))
A.discard(x_val)
print("Set A after discard:", A)
```

## Q5. Dictionaries
---

```python
my_dict = {
    "name": "Kovid Kumar",
    "roll_no": "1024170353",
    "branch": "COPC",
    "age": 20,
    "city": "Ambala, Haryana"
}
print("Original dictionary:", my_dict)
```

**i.** Rename key `city` to `location` without hand-recreating the dict

```python
my_dict["location"] = my_dict.pop("city")
print("After renaming 'city' to 'location':", my_dict)
```

**ii.** Add a new key `cgpa`

```python
my_dict["cgpa"] = 8.13
print("After adding 'cgpa':", my_dict)
```

**iii.** Update the value of `age` by increasing it by 1

```python
my_dict["age"] += 1
print("After incrementing age:", my_dict)
```

**iv.** Delete `branch` two ways, in two separate copies — `pop()` and `del`

`pop()` removes the key AND returns its value (usable in an expression), while
`del` simply removes the key-value pair and returns nothing.

```python
dict_pop = my_dict.copy()
popped_value = dict_pop.pop("branch")
print("dict after pop('branch'):", dict_pop)
print("pop() returned the removed value:", popped_value)

dict_del = my_dict.copy()
del dict_del["branch"]
print("dict after del dict_del['branch']:", dict_del)
```

**v.** Iterate over the dictionary using `.items()`

```python
for key, value in my_dict.items():
    print(f"{key} -> {value}")
```

**vi.** Safely check for a non-existent `email` key before accessing it

```python
if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("'email' key not found - no email on record")
```

**vii.** Merge with a friend's dictionary using `{**dict1, **dict2}`

When both dictionaries share a key, the value from the SECOND dict (`friend_dict`)
wins, since it is unpacked last and overwrites the earlier keys.

```python
friend_dict = {
    "name": "Rohan Verma",
    "roll_no": "1024170399",
    "branch": "COE",
    "age": 21,
    "city": "Chandigarh"
}
merged_dict = {**my_dict, **friend_dict}
print("Merged dictionary:", merged_dict)
```

**viii.** Dictionary comprehension — keep only string-valued pairs

```python
string_only_dict = {k: v for k, v in my_dict.items() if isinstance(v, str)}
print("Only string-valued key-value pairs:", string_only_dict)
```
